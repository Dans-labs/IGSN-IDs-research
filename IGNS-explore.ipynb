```python
import requests

def print_catalog_igsn(client_id, client_name, size=5):
    q =  f"https://api.datacite.org/dois?client-type=igsnCatalog&client-id={client_id}&page[size]={size}"
    response = requests.get(q).json()
    print(f'{"-"*10} IGSN Catalog: {client_name} ({client_id}) {"-"*10}')
    for pid in response["data"]:
        # print(f'IGSN: {pid["id"]},     title:  {pid["attributes"]["titles"][0]["title"]}    publisher: {pid["attributes"]["publisher"]}  ')
        print(pid["id"], f"commons: https://commons.datacite.org/doi.org/{pid['id']}"  f" landing page: https://doi.org/{pid['id']}" )
```

```python
# IGSN Catalogs
q =  "https://api.datacite.org/clients?client-type=igsnCatalog"
response = requests.get(q).json()
for client in response["data"]:
    print(f'CLIENT id: {client["id"]},     name:  {client["attributes"]["name"]}    prefix: {client["relationships"]["prefixes"]["data"][0]["id"]}  ')
    # pprint(client)
```

```python

print_catalog_igsn(client_id="scpr.uoadij", client_name="4TU.ResearchData - IGSN Catalog")
print_catalog_igsn(client_id="arch.igsn", client_name="Archaeological Map of the Czech Republic - IGSN Catalog")


```
