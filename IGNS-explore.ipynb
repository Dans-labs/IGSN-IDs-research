```python
import csv
import requests

def catalog_igsns_iterable(client_id, client_name, size=5):
    # yeields an iterable of IGSNs for a given client (catalog) from the DataCite API
    q =  f"https://api.datacite.org/dois?client-type=igsnCatalog&client-id={client_id}&page[size]={size}"
    response = requests.get(q).json()
    data = response["data"]
    if len(data) == 0:
        print(f"No IGSNs found in Catalog: {client_id}")
        yield [client_id, client_name, None, None, None]
    else:
        for igsn in data:
            yield [client_id, client_name, igsn['id'], f"https://doi.org/{igsn['id']}", f"https://commons.datacite.org/doi.org/{igsn['id']}"]
```

```python
# write IGSN Catalogs and some of their IGSN IDs to igsn_catalogs_ids.csv

# IGSN Catalogs from DataCite API
datacite_igsn_cats =  "https://api.datacite.org/clients?client-type=igsnCatalog"
datacite_igsn_cats_resp = requests.get(datacite_igsn_cats).json()
# CSV
with open('igsn_catalogs_ids.csv', 'w') as csv_writer:
    csv_writer = csv.writer(csv_writer)
    csv_writer.writerow(['catalog_id', 'catalog_name', 'igsn', 'landing_page', 'datacite_commons_page ']) # headers
    for catalog in datacite_igsn_cats_resp["data"]:
        print(f'Catalog id: {catalog["id"]},     name:  {catalog["attributes"]["name"]}    prefix: {catalog["relationships"]["prefixes"]["data"][0]["id"]}')
        csv_writer.writerows(catalog_igsns_iterable(client_id=catalog["id"], client_name=catalog["attributes"]["name"]))
```
