```python
import csv
import json
import requests
from pathlib import Path

json_dir = Path("IGSN-DataCite-metadata")

with open('igsn_catalogs_ids.csv', 'r') as csv_reader:
    reader = csv.DictReader(csv_reader)
    for row in reader:
        if row['igsn']:
            print(row['igsn'])
            json_url = f"https://api.datacite.org/application/vnd.datacite.datacite+json/{row['igsn']}"
            response = requests.get(json_url)
            if response.status_code == 200:
                data = response.json()
                json_path = json_dir / f"{row['igsn']}.json"
                json_path.parent.mkdir(parents=True, exist_ok=True)
                with open(json_path, 'w') as json_file:
                    json.dump(data, json_file, indent=4)
```
